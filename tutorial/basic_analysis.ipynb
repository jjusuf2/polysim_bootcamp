```python
import numpy as np
from matplotlib import pyplot as plt

from polychrom.hdf5_format import list_URIs, load_URI
```

```python
# First, find the files in the output folder
output_folder = '/mnt/md1/jjusuf/polysim/outputs/trajectory_LEF_EP_absquant_region_bootcamp_version'
files = list_URIs(output_folder)
print(f'{len(files)} saved blocks found')
```

```python
# Take a look at the objects in this list: each one represents a saved block
files
```

```python
# We can open one of the files using load_URI
save_block_num = 56711  # the 56,711th saved block
data = load_URI(files[save_block_num])

# The data inside dictionary with the following keys
data.keys()
# only 'SMCs' and 'pos' are important
```

```python
# The 'pos' entry contains the xyz positions of all monomers
# The ith row is monomer i, and the columns represent x,y,z.
# It is an Nx3 numpy array
print(data['pos'])
```

```python
# The 'SMCs' entry shows you the locations of the cohesins (aka LEFs)
# Each row represents the [left_pos, right_pos] of a cohesin
# It is not in any particular order
data['SMCs']
```
