```python
from polychrom.hdf5_format import list_URIs, load_URI
```

```python
# load my data
files_no_LE_no_EP = list_URIs('/mnt/md1/jjusuf/polysim/outputs/calibration3D/trajectory0007_npoly70000_nchr1_dens0.3_noLEF_dt40')
print(f'{len(files_no_LE_no_EP)} blocks found')
```

```python
xyz = load_URI(files_no_LE_no_EP[0])['pos']
xyz_subset = xyz[:1000:]
```

```python
import py3Dmol
import matplotlib as mpl

# one "atom" per monomer; the element symbol is arbitrary
n = len(xyz_subset)
xyz_block = f"{n}\nmonomers\n" + "\n".join(
    f"C {x:.3f} {y:.3f} {z:.3f}" for x, y, z in xyz_subset
)

view = py3Dmol.view(width=600, height=500)
view.addModel(xyz_block, 'xyz')

# ball model, colored blue -> red along the chain
cmap = mpl.colormaps['coolwarm']
for i in range(n):
    color = mpl.colors.to_hex(cmap(i / (n - 1)))
    view.setStyle({'index': i}, {'sphere': {'radius': 0.5, 'color': color}})

view.setBackgroundColor('white')
view.zoomTo()
view.show()
```
